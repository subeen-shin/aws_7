## 클래스
* 필드 선언은 무조건 생성자에게만 가능 => x
* 접근제어자가 없음
* 클래스의 함수나 생성자는 반드시 첫번째 매개변수로 `self`를 넘겨주어야 함
```
class 클래스(부모클래스):

    변수 = 초기값 # 클래스 변수

    # 생성자 : 보통 필드들을 선언 후에 초기화
    def __init__(self, 변수1, 변수2, ...):
        self.필드1 = 변수1 # 필드1은 자바의 인스턴스 변수(필드)

    # toString
    def __str__(self):
        return 필드를 포함한 문자열

    # 필요한기능을 함수로 만듬
    
```

````python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"이름 : {self.name}, 나이 :  {self.age}"

    # 기능 : 학생 정보를 콘솔에 출력하는 기능(print)
    # 이름 : 홍길동
    # 나이 : 17살

    def print(self): # self의 역할은 선언할 때 쓰이지 않고 호출할때만 쓰임.. 호출할때는 안적어도됨
        print(f"이름 : {self.name}")
        print(f"나이 : {self.age}")
        
std1 = Student("홍길동", 17)
print(std1)
std1.print()
````

````python
from dataclasses import dataclass, field, asdict

@dataclass # 생성자를 자동으로 추가, __str__도 자동으로 추가
class Student2:
    name:str # 클래스 변수를 필드로 만듬
    age:int
    # 전화번호. 집전화, 폰번호
    numbs:list[str] = field(default_factory=list) # []로 초기화 하지 않고 field(default_factory=list)로 빈 리스트 초기화
   
    def print(self): # self의 역할은 선언할 때 쓰이지 않고 호출할때만 쓰임.. 호출할때는 안적어도됨
        print(f"이름 : {self.name}")
        print(f"나이 : {self.age}살")
        
std2 = Student2("홍길동", 17)
std3 = Student2("임꺽정", 18)
print(std2)
print(std3)
print(asdict(std2)) # @dataclass를 이용하여 선언한 클래스의 객체를 dict로 변환

````

````python
from dataclasses import dataclass, field

# BankAccount 클래스를 만드세요.
# 계좌 정보 : owner(소유자), balance(잔액)
# 기능 : 입금(deposit)
    # 입금액이 0이하면 "금액은 0보다 커야합니다" 출력
# 기능 : 출금(withdraw)
    # 출금액이 0이하이면 "금액은 0보다 커야 합니다" 출력
    # 잔액보다 출금액이 크면 "잔액이 부족합니다"를 출력
# 기능 : 잔액확인(get_balace)
# 기능 : 거래내역 조회(print_history)
    # 입금 : 3000원
    # 출금 : 2000원
@dataclass
class BankAccount:
    owner:str
    balance:int
    # 거래 내역
    history:list[str] = field(default_factory=list)

    def deposit (self, 입금액):
        if 입금액 <= 0:
            print("금액은 0보다 커야합니다")
        else:
            self.balance += 입금액
            content = f"입금 : {입금액}원"
            # 기록을 저장
            self.history.append(content)
            # 콘솔 출력
            print(content)
    def get_balance(self):
        return self.balance

    def print_history(self):
        print("입출금 내역")
        for item in self.history:
            print(item)
            
    def withdraw (self, 출금액):
        if 출금액 <= 0:
            print("금액은 0보다 커야합니다")
        elif 출금액 > self.balance:
            print("잔액이 부족합니다")
           
        else:
            self.balance -= 출금액
            content = f"출금 : {출금액}원"
            # 기록을 저장
            self.history.append(content)
            # 콘솔 출력
            print(content)
    
    

#    def __init__(self, owner, balace):
#        self.owner = owner
#        self.balance = balance

#    def __str__(self):
#        return f"소유자 : {self.owner}, 잔액 : {self.balace}"

#    def deposit (self, amount):
#        if amount <= 0:
#            print("금액은 0보다 커야합니다")
#           return 
#            self.balace += amount 
        
#     def withdraw (self, amount):
#        if amount <= 0:
#            print("금액은 0보다 커야합니다")

#    def get_balace (self, amount):
        

    
acc = BankAccount("홍길동", 10000)
acc.deposit(5000)
acc.deposit(-5000)
acc.withdraw(30000)

print(f"잔액 : {acc.get_balance()}")
acc.print_history()
````
