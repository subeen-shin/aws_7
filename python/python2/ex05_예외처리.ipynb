### 예외처리
* 프로그램을 안정적으로 실행되기 하기 위해
* 실행 중에 발생할 수 있는 문제를 처리
```
try:
    #예외가 발생할 수 있는 코드들
except 에러 as e:
    # 예외처리
except 에러2 as e:
    # 예외처리
else: # try문에서 예외가 발생 안하면 실행
    # 실행 코드
finally: # 예외 발생 여부와 상관 없이 무조건 실행 #try 또는 except에서 return을 해도 반드시 실행
    # 실행 코드
```


````python
num1, num2 = 1, 0
try:
    print(num1 / num2)
except ZeroDivisionError as e:
     print("0으로 나눌 수 없습니다.")
     print(e)
````

### 예외 발생시키기
* 예외가 발생하는 코드는 아니지만 상황에 따라 예외를 발생시켜야 할 때 사용
* 은행에서 출금할 때 출금액이 0이면 문제가 됨. 0 자체는 예외가 발생하지 않음
* `raise`를 이용하여 예외를 발생시킴


````python
def deposit(balance, amount):
    if amount <= 0:
        raise ValueError("입금액은 0보다 커야합니다.")
    print(f"입금액 : {amount}")
    return balance + amount

balance = 1000
try:
    balance = deposit(balance, 100)
    balance = deposit(balance, -100)
    balance = deposit(balance, 1000)
except ValueError as e:
    print(e)

print("----------")
balance = 100
for amount in (100, -100, 1000):
    try:
        balance = deposit(balance, amount)
    except ValueError as e:
        print(e)
print(f"잔액 : {balance}")
````
