### 문제4
* 다음 클래스를 만드세요.
* 필드(인스턴스 변수) : 리스트
* 메서드(기능, 함수) : 리스트_출력해(필드인 리스트를 출력하는 함수)
* 생성자 : 밖에서 리스트를 넘겨주면 리스트를 가져와서 필드에 저장
* 문법
```
class 클래스명:
    def __init__(self, 변수들): # 생성자
        생성자 구현
        self.필드명 = 변수값1
        
    def 메서드명(self, 변수들):
        메서드 구현
        
```


* 파이썬은 오버로딩 기능을 제공하지 않음
    * 오버로딩 : 같은 이름의 함수, 메서드, 생성자가 있는 것
* 파이썬은 매개변수의 기본값을 이용하여 오버로딩 기능을 흉내낼 수 있다
* 

````python
class Num :
    def __init__(self, numbers=None): #numbers=None : list값을 넣어주지 않으면 None이 자동으로 들어감
        if numbers is None: # 자바는 list == null 파이썬에서 None은 is또는 is not 이용
            self.numbers = []
        else:
         # 파이썬에서 필드는 미리 선언하지 않고 사용해도 됨
            self.numbers = numbers
       
    def print(self):
        for 값 in self.numbers:
            print(값)
       # print(f"숫자 : {self.numbers}")
````

### 문제6
* 리턴값이 2개인 함수를 만들고 호출
  

````python
def 리스트_최대값_최소값(Num):
    최소값 = min(Num)
    최대값 = max(Num)
    return 최소값, 최대값

리스트 = [10, 8, 12, 100]
결과 = 리스트_최대값_최소값(리스트) # 최소값과 최대값을 결과라는 변수 하나에 저장
print(결과) # 여러개를 return 하는 경우 타입은 튜플
print(f"최소값 : {결과[0]}, 최대값 : {결과[1]}")

최소값, 최대값 = 리스트_최대값_최소값(리스트) # 최소값과 최대값을  최소값과 최대값이라는 변수 2개에 각각 저장
print(f"최소값 : {결과[0]}, 최대값 : {결과[1]}")
````

### 문제8
* 다음 키워드 인자 코드를 보고 결과를 예상해보세요.

```
def order(menu, size="M", count=1):
    print(f"{menu} {size} 사이즈{count}잔")

order("아메리카노")
order("라떼", "L")
order("라떼", count=3)
order("count=2", menu="모카")
order(menu="녹차", "5")
order("아메리카노", size=""L, size="M")

```

````python
def order(menu, size="M", count=1):
    print(f"{menu} {size} 사이즈{count}잔")

order("아메리카노")
order("라떼", "L")
order("라떼", count=3)
order("count=2", menu="모카")
order(menu="녹차", "S")
order("아메리카노", size=""L, size="M")
````
