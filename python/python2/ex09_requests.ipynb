```python
# requests는 js의 fetch처럼 url을 입력하면 해당 결과를 가져오는 기능을 가짐
import requests as re

# 최대 5초 대기
result = re.get("https://jsonplaceholder.typicode.com/posts", timeout=5)
print(result.status_code)
posts = result.json()
```

```python
# posts에 있는 게시글 정보를 콘솔에 반복문으로 출력하는 코드를 작성
# userId : title
# 1 : sunt aut facere repellat provident occaecati excepturi optio reprehenderit 

for post in posts:
    print(f"{post["userId"]} : {post.get("title")}")
```

```python
# posts를 pandas의 DataFrame으로 변환
import pandas as pd

df = pd.DataFrame(posts)
print(df.head(5))
```

```python
# 만들어진 df에 결측치가 있는지 확인
df.isna().sum()
```

## timeout 처리

```python
try:
    result = re.get("https://jsonplaceholder.typicode.com/posts", timeout=0.001)
except re.Timeout:
    print("요청 시간 초과")
```

## 오류 처리(404)

```python
result = re.get("https://jsonplaceholder.typicode.com/posts/9999", timeout=5)

try:
    result.raise_for_status()
except re.HTTPError as e:
    print(f"Http 오류 : {e}")
```

## requets 사용 기본 형태

```python
BASE_URL = "https://jsonplaceholder.typicode.com"

# 게시글을 가져오는 함수. post는 post방식이 아닌 게시글을 의미
def fetch_post(post_id:int) -> dict|None:
    try:
        result = re.get(f"{BASE_URL}/posts/{post_id}", timeout=5)
        result.raise_for_status()
        return result.json()
    except re.Timeout:
        print("요청 시간 초과")
    except re.RequestException as e:
        print(f"요청 실패 : {e}")
    return None

post = fetch_post(101)
if post is not None:
    print(post)
```
