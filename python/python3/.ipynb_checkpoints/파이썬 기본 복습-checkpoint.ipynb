## 조건문

```python
# if문
# 숫자가 음수, 양수, 0인지를 판별하는 예제
num = 10
if num > 0 :
    print(f"{num}는 양수")
elif num < 0 :
    print(f"{num}는 양수")
else : 
    print(0)
```

# match문(자바의 switch문 역할을 함)
# month가 12, 1, 2이면 겨울, 3,4,5이면 봄, 6, 7, 8이면 여름, 9,10,11이면 겨울
month = 1
match month:
    case 12 | 1 | 2:
        print(f"{month}는 겨울")
    case 3 | 4 | 5:
        print(f"{month}는 겨울")
    case 6 | 7 | 8:
        print(f"{month}는 겨울")
    case 9 | 10 | 11:
        print(f"{month}는 겨울")
    case _:
        print(f"{month}는 잘못된 월입니다")

# 딕셔너리 : dict

```python
# key와 value로 이루어짐(자바의 Map)
# key는 중복 불가능
# 순서를 보장
# 수정 가능
# {}를 이용 {키 : 값}
user = { "id" : 1, "name" : "홍길동", "age" : 21, "id" : 2}
print(user)
# 새로운 키 추가
user["nick"] = "홍길동"
print(user)
# 키 삭제
del user["age"]
print(user)
# 키값 가져오기
# print(user["address"]) #address 키가 없어서 에러 발생(KeyError)
print(user.get("address")) # address 키가 없어서 None을 반환
```

```python
name = "홍길동"
print(name)
print(name[0])

# 이름(한글)을 주면 성을 알려주는 함수를 만들고 호출
# 단성은 무조건 1글자라고 가정. 이름은 반드시 성을 첫글자에 포함
# 함수 이름 : get_family_name

def 성을_알려줘(이름):
    return 이름[0]

print(성을_알려줘("홍길동"))
```

```python
# 이름을 주면 이름의 길이를 알려주는 함수

def namelen(name):
    return len(name)    

print(namelen("홍길동"))

```

```python
# 구매 제품 가격과 구매 제품수량이 주어지면 총 가격을 알려주는 함수

def total(price, amount):
    return price * amount

price = 5000
amount = 2
print(total(price, amount))
```

```python
# 만 19세는 성인이고, 미만은 청소년일때
# 만 나이를 이용해서 성인인지 청소년인지 확인하는 코드
# 이 때 필요한 함수를 만들어 사용

# 만 19세는 성인이고, 미만은 청소년일때
# 만 나이를 이용해서 성인인지 청소년인지 확인하는 코드를 작성하세요.
# 이 때 필요한 함수를 만들어 사용

나이 = 20

# if 나이 >= 19:
#     print(f"{나이}살은 성인입니다.")
# else:
#     print(f"{나이}살은 청소년입니다.")

def 나이_확인1(나이):
    if 나이 >= 19:
        print(f"{나이}살은 성인입니다.")
    else:
        print(f"{나이}살은 청소년입니다.")
        
나이_확인1(나이)

def 나이_확인2(나이2):
    if 나이2 >= 19:
        return "성인"
    else:
        return "청소년"

print(f"{나이}살은 {나이_확인2(나이)}입니다.")

def 성인입니까(나이3):
    if 나이3 >= 19:
        return True
    else:
        return False

성인입니까(나이) # True

if 성인입니까(나이):
    print(f"{나이}살은 성인입니다.")
else:
    print(f"{나이}살은 청소년입니다.")
```

```python
# 성적(0~100)이 커트라인(예:60) 이상이면 합격, 아니면 불합격을 출력하는 코드 작성
# 함수 이용 / 커트라인 최소 60

score = 90
cutline = 60

def check_pass(score, cutline=60): #함수의 기본값을 적용
    if score >= cutline:
        return "합격"
    else:
        return "불합격"

print(f"{score}점은 {check_pass(score, cutline)}") #함수 호출할 때 기본 값이 있는 매개 변수는 생략해도 됨
```

```python
# 이름(name), 나이(age), 주소(address)를 주면 딕셔너리로 만들어주는 함수를 만들고
# 다음처럼 호출했을 때 실행되도록 작업

def to_dict(age, name="익명", address="없음"):
    return {
        "name" : name, 
        "age" : age, 
        "address" : address
    }
    
    
user1 = to_dict(21, "홍길동", "서울시")
# user1 = { "name" : "홍길동", "age" : 21,  "address" : "서울시" }
print(user1)

user2 = to_dict(21, "홍길동")  # 홍길동, 21, 없음
print(user2)
user3 = to_dict(21) # 익명, 21, 없음 
print(user3)
```
