```python
import requests

res = requests.get("https://jsonplaceholder.typicode.com/posts/1", timeout=5)
print(res.status_code)        # 200
post = res.json() # 요청 결과를 json으로
print(type(post)) #
print(post["id"], post["title"])
```

```python
res = requests.get(
    "https://jsonplaceholder.typicode.com/posts",
    params={"userId": 1}, # ?userId=1&a=b 형태로 변환
    timeout=5,
)
print(res.url)                # 실제로 호출된 주소
print([p["title"] for p in res.json()[:3]])
```
