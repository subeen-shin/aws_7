```python
import pandas as pd

# 엑셀 표를 만드는데 열 이름에 name, score, class가 들어감
df = pd.DataFrame([
    {"name": "김철수", "score": 92, "class": "A"},
    {"name": "이영희", "score": 75, "class": "B"},
    {"name": "박민수", "score": 88, "class": "A"},
    {"name": "최지우", "score": 66, "class": "B"},
])

print(df)
# shape : 몇행(데이터 개수) 몇열(속성 개수)인지
print(df.shape)        # (4, 3)
# head : 전체 중에 앞부분 2줄
print(df.head(2))
# info : 컬럼 별 타입, 빈 값 개수
df.info()             
```

```python
print(df["name"]) # name 열만 조회
print(df[["name", "score"]]) # name과 score 열만 조회
print(df["score"].mean(), df["score"].max()) # score의 최소값, 최대값

high = df[df["score"] >= 80] # score가 80이상인 행들의 name, score, class 조회
a_high = df[(df["class"] == "A") & (df["score"] >= 90)] # 반이 A class이고 score가 90이상인 행들을 조회

print(df.sort_values("score", ascending=False)) # score으로 조회 (내립차순)
```

```python
# 새칼럼 추가
# 'passed' 컬럼 추가. 성적이 60이상이면 passed
df["passed"] = df["score"] >= 70
print(df)
```

```python
# 그룹
by_class = df.groupby("class")["score"].mean()
print(by_class)
print(by_class.to_dict())
```

```python
# 저장 및 불러오기
# df.to_csv("파일명", index=False, encoding="utf-8-sig")
# index=False, 인덱스를 저장하지 않음.
# encoding="utf-8-sig", 저장할 때 한글 깨짐을 방지
df.to_csv("scores.csv", index=False, encoding="utf-8-sig")
```

```python
# csv파일 불러오기
df2 = pd.read_csv("scores.csv")
print(df2)
```
